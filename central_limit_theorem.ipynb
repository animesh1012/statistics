# Central Limit Theorem

```python
import numpy as np
```

## 1. Population Initialization

```python
##Taking random number between 1 to 100 (100 not included)
population = np.random.randint(1,100,10000)
pop_mean = np.mean(population)
print("Shape of Population:",population.shape)
print("Population:",population)
print("Population Mean:",pop_mean)
```

## 2. Sampling
- Randomly selecting `40 observation` from population
- Creating 500 sample having `40 observation`

```python
mean_of_sample = []
for i in range(500):
    X1 = np.random.choice(population,size=40)
    mean_of_sample.append(np.mean(X1))
mean_of_sample = np.array(mean_of_sample)
mean_of_sample
```

## 3. Plotting Histogram of Sample Mean

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.histplot(mean_of_sample, bins=10, kde=True, color='orange', edgecolor='red')
plt.title("Histogram of Sample Mean")
plt.xlabel('Values')
plt.ylabel('Density')
plt.show()
```

```python
## Mean value of Sample mean
mean_of_sample_mean = np.mean(mean_of_sample)
print("Mean of Sample Mean:",mean_of_sample_mean)
##Recalling population mean
print("Population Mean:",pop_mean)
```

### Insight
- Above Result shows that, Mean of Sample mean is approximately equal to Population Mean.
- Histogram of Sample mean shows that its a normal distribution.
